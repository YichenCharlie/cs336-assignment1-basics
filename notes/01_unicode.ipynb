2.2 Unicode Encodings
a, oh i know , when we use the utf-8 , one byte to one char, like hello , in utf-8 it just represent by 5 bytes, but in utf-16 , it just represent up to 10 bytes. so when we use the utf-8 , we can use less bytes to express string.
GPT_explain:UTF-8 uses one byte per ASCII character, so “hello” takes 5 bytes in UTF-8, compared with 10 in UTF-16 and 20 in UTF-32, excluding byte-order marks. For text containing mostly ASCII characters, UTF-8 therefore gives a byte-level tokenizer a shorter sequence to process.

```python
def decode_utf8_b_to_s_wrong(bytestring):
    return "".join([bytes([b]).decode("utf-8") for b in bytestring])

decode_utf8_b_to_s_wrong("你".encode("utf-8"))
```

```python
b, when i use the example "你", it will go wrong, letters can be transfer to character one by one , but when we need to represent to some special char like chinese words , it need 3 bytes to express one word, so when we use loop to decode the string ,
 one single byte can not be decoded
GPT:The function works for "hello" because each letter is encoded as a single UTF-8 byte, but it fails on "你".encode("utf-8") because it tries to decode each of the character’s three bytes separately.
```
